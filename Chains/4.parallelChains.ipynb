```python
from langchain_ollama import ChatOllama
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnableParallel
```

```python
model1 = ChatOllama(model="gpt-oss:120b-cloud")
model2 = ChatOllama(model="gpt-oss:120b-cloud")

prompt1 = PromptTemplate(
    template="Generate short and simple note from the following topic {topic}",
    input_variables=['topic']
)

prompt2 = PromptTemplate(
    template="Generate 5 short question answer from the following text {text}",
    input_variables=['text']
)

prompt3 = PromptTemplate(
    template="Merge the provided notes and question answers into single document {notes},{qa} ",
    input_variables=['notes','qa']
)


parser = StrOutputParser()

runnable_chain = RunnableParallel(
    {
        'notes':prompt1|model1|parser,
        'qa':prompt2|model2|parser
    }
)

merge_chain = prompt3|model1|parser

final_chain = runnable_chain|merge_chain

text = """ Support Vector Machine (SVM) is a supervised machine learning algorithm used for classification and regression tasks.
It works by finding the optimal hyperplane that best separates different classes of data.
The main objective of SVM is to maximize the margin between the hyperplane and the nearest data points.
These nearest data points are called support vectors, which influence the position of the hyperplane.
SVM is effective in high-dimensional datasets and can handle complex classification problems.
It can perform both linear and non-linear classification using a technique called the kernel trick.
Common kernel functions include Linear, Polynomial, Radial Basis Function (RBF), and Sigmoid kernels.
SVM is less prone to overfitting, especially when the number of features is large.
It is widely used in applications such as text classification, image recognition, and bioinformatics.
Although SVM provides high accuracy, it can be computationally expensive for very large datasets.
Provide your feedback on BizChat
Explain the kernel trick in detail.How does SVM handle non-linear data?What are the limitations of SVM?
FilesFiles"""

response = final_chain.invoke(text)

print(response)
```
