```python
from langchain_ollama import ChatOllama
from typing import TypedDict,Annotated,Optional,Literal

from pydantic import BaseModel, Field
```

```python
model = ChatOllama(model="llama3.2")

class Schema(BaseModel):
    keythemes: list[str] = Field(description="Write down the key themes discussed in the review in a list")
    summary:str = Field(description="write a brief summary of the review")
    sentiment:Literal["pos","neg"]= Field(description="Return sentiment of the review")
    pros:Optional[list[str]]= Field(default= None,description="Write down all the pros in the list")
    cons:Optional[list[str]]= Field(default=None,description="Write down all the cons in the list")
    name:Optional[str] = Field(default=None,description="Write the name of the reviewer if found in the review.Generally found at the end of the review")

structured_model = model.with_structured_output(Schema,strict =True)

prompt = """
Write a detailed and engaging article about Google Pixel smartphones in a continuous paragraph-based format rather than using bullet points or numbered lists. Explain how Google Pixel evolved from Google's vision of creating a pure Android experience and how it has become one of the most innovative smartphone brands in the industry. Discuss the design philosophy of Pixel phones, including their premium build quality, distinctive camera bar design, high-quality OLED displays, and focus on simplicity and usability.
Describe the hardware capabilities of Pixel devices, with special emphasis on Google's Tensor processors and how they are optimized for artificial intelligence and machine learning tasks. Explain how Tensor chips differ from traditional smartphone processors and how they enhance the overall user experience through smarter performance, improved voice recognition, and efficient power management.
Dedicate significant attention to the camera system, one of the most recognized features of Google Pixel phones. Explain Google's approach to computational photography, including technologies such as HDR+, Night Sight, Portrait Mode, Super Res Zoom, Astrophotography Mode, and AI-powered image enhancement. Discuss how Pixel devices use software and machine learning to produce professional-quality photographs and videos in a variety of lighting conditions.

Explore the software experience offered by Pixel phones, focusing on the clean Android interface, exclusive Pixel features, long-term software support, regular security updates, and deep integration with Google services. Describe how features such as Live Translate, Call Screening, Circle to Search, Voice Typing, Gemini AI integration, and Magic Editor enhance productivity and convenience in everyday life.

Analyze the role of artificial intelligence within the Pixel ecosystem and explain how Google is positioning Pixel phones as AI-first devices. Discuss on-device AI processing, contextual assistance, real-time language translation, intelligent photo editing, voice interaction, and personalized recommendations. Explain how these capabilities differentiate Pixel devices from many competitors in the smartphone market.

Examine security and privacy features, including Google's security architecture, encryption technologies, secure authentication methods, and privacy-focused controls. Discuss how Google protects user data while maintaining a seamless user experience.

Evaluate battery performance, fast charging capabilities, wireless charging support, connectivity features such as 5G, Wi-Fi, Bluetooth, and NFC, and how these contribute to a modern smartphone experience. Explain the integration of Pixel devices with Google's wider ecosystem, including Google Photos, Google Drive, Gmail, Google Maps, Google Workspace, and smart home - Kunal Borse """


response = structured_model.invoke(prompt)
print(response)
```
