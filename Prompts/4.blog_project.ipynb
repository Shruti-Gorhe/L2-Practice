```python
from langchain_core.prompts import ChatPromptTemplate, SystemMessagePromptTemplate, HumanMessagePromptTemplate, AIMessagePromptTemplate
from langchain_google_genai import ChatGoogleGenerativeAI
from dotenv import load_dotenv
import os
load_dotenv()
```

```python
chat_model = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash")

print("Blog post generator")
print("Provide ideas or topics for the blog post. You can also provide a title and keywords.Type exit to finish")

topic = input("Enter your topic: ")

chat_prompt = ChatPromptTemplate.from_messages([
    SystemMessagePromptTemplate.from_template("You are a professional blog writer.Help generte informative, engaging and well-structured blog posts based on the provided {topic}, title, and keywords."),
    HumanMessagePromptTemplate.from_template("Write a detailed blog post on the topic: {topic}. The blog post should be")
])

#initialize the chat history to store in list
chat_history = []
while True:
    if topic.lower() == "exit":
        break
    #generate the blog post using the chat model and prompt
    response = chat_model.invoke(chat_prompt.format_messages(topic=topic))
    print("Generated Blog Post:\n")
    print(response.content)
    #store the generated blog post in the chat history
    chat_history.append({"topic": topic, "blog_post": response.content})
    #ask for another topic
    topic = input("\nEnter another topic or type exit to finish: ")


```
