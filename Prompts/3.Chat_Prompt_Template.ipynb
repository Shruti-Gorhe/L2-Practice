```python
from langchain_core.prompts import ChatPromptTemplate, SystemMessagePromptTemplate, HumanMessagePromptTemplate, AIMessagePromptTemplate

chat_prompt = ChatPromptTemplate.from_messages(
    [
        SystemMessagePromptTemplate.from_template("You are a helpful assistant that provides information about {subject}."),
        HumanMessagePromptTemplate.from_template("Write a short fun fact about {subject}."),
        AIMessagePromptTemplate.from_template("Sure! Did you know that the term 'Artificial Intelligence' was first coined in 1956 at a conference at Dartmouth College?"),
    ]
)

prompt= chat_prompt.format_prompt(subject="quantum computing").to_messages()
```
